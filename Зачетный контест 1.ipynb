```python
def main():
    n, m = map(int, input().split())
    
    table = []
    
    for i in range(n):
        row = list(input().strip())
        table.append(row)
    
    # Выводим считанную таблицу для проверки
    print(f"Размеры таблицы: {n} x {m}")
    print("Содержимое таблицы:")
    for row in table:
        print(''.join(row))
    
    # Здесь можно работать с массивом table
    # table[i][j] - символ в i-й строке и j-м столбце
    # table[i][j] == '#' - занятая клетка
    # table[i][j] == '.' - свободная клетка

    ans = 0
    for i in range(n):
        pred = table[i][0]
        for j in range(1, m):
            if table[i][j] == pred == ".":
                ans += 1
            pred = table[i][j]


    print(ans)
    
    return


if __name__ == '__main__':
    main()
```

```python
def main():

    n, c, r = map(int, input().split())

    
    cs = list(map(int, input().split()))
    c = set(key for key in cs)

    rs = list(map(int, input().split()))
    r = set(key for key in rs)

    ans = 0
    for man in range(1, n+1):
        if not man in c:
            ans += 1
        else:
            if man in r:
                ans += 1
                r.remove(man)
            else:
                if man-1 in r:
                    ans += 1
                    r.remove(man-1)
                elif man+1 in r and not man in c:
                    ans += 1
                    r.remove(man+1)

    print(ans)

if __name__ == '__main__':
    main()
```
